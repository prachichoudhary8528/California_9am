```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x_train,y_train)
# model developed so we have coefficient and intercept
```

```python
x_train.values[0],x.columns,y_test.values
```

```python
# linear regressionn equation
# lr= b_0 + b1*MedInc + b2*HouseAge + b3*AveRooms + b4*Population + b5*AveOccup + b6*Latitude
# lr= b_0 + b1*1.7875  + b2* 24.  + b3*3.5335277 + b4* 1202. + b5*3.50437318 + b6* 37.96 
# lr=1.6174822629220276 + 4.42489549e-01*1.7875 + 1.71325075e-02*24. + -3.76571837e-02 *3.5335277 
#       + 1.86502053e-05* 1202. + -6.78419200e-03*3.50437318 + -4.36663247e-02* 37.96 
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
import numpy as np
lr.coef_*(x_train.values[0])
y_pred0=lr.intercept_+(sum(lr.coef_*(x_test.values[0])))
print(y_pred0,y_test.values[0])

y_pred1=lr.intercept_+sum(lr.coef_*(x_test.values[1]))
print(y_pred1,y_test.values[1])

#like this there will 4128 predictions and 4128 errors 
# then find error,mse,rmse
```

```python
y_test_p0= lr.intercept_ + np.sum(lr.coef_*x_test.values[0])
y_test_p1= lr.intercept_ + np.sum(lr.coef_*x_test.values[1])
y_test_p2= lr.intercept_ + np.sum(lr.coef_*x_test.values[2])
y_test_p3= lr.intercept_ + np.sum(lr.coef_*x_test.values[3])
```

```python
print(y_test_p0,y_test_p1,y_test_p2,y_test_p3)
```

```python
y_test_p= lr.intercept_ + np.sum(lr.coef_*x_test.values,axis=1)
y_test_p
```

```python
df1=x_test.copy()
df1["y_actual"]=y_test
df1["y_pred"]=y_test_p
df1["error"]=y_test-y_test_p
df1["sq_error"]=np.square(y_test-y_test_p)
sq_error=np.square(y_test-y_test_p)
df1["sqrt"]=np.sqrt(sq_error)
df1
```

```python
y_pred=lr.predict(x_test)
mse=lr.mean_squared_error(y_test,y_test_p)
rmse=lr.root_mean_squared_error(y_test,y_test_p)
r2=lr.r2_score(y_test,y_pred)
rmse,mse
```

```python
# i want to save a model
# where i want to save
# what name i want to save
# which extension you want to use
import pickle
file=open('california.pkl','wb')
pickle.dump(lr,file)
```
